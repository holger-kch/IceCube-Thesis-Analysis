# All variables for one DOM hit (pulse-level parquet)

Each row in the `data_parquet/` files is **one reconstructed pulse** (one
DOM hit) from `SplitInIcePulses`. This notebook prints every column for a
single hit so you can see what the pulse-level data contains.

Source: `../data_parquet/`

```python
import pandas as pd
from pathlib import Path

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

PARQUET_DIR = Path('../data_parquet')
PARQUET = PARQUET_DIR / 'data_SplitInIcePulses_stopped.parquet'

df = pd.read_parquet(PARQUET)
print(f'file:   {PARQUET.name}')
print(f'rows:   {len(df):,} pulses')
print(f'events: {df["event_no"].nunique():,}')
print(f'columns ({len(df.columns)}): {list(df.columns)}')
```

## Pick one hit and print all its variables
Change `ROW` to look at a different pulse.

```python
ROW = 0   # any row index

hit = df.iloc[ROW]
print(f'DOM hit at parquet row {ROW} (event_no = {int(hit["event_no"])}):')
print()
# transpose so every variable is on its own line
print(hit.to_frame(name='value'))
```

## Same hit as a labelled list (with dtypes)

```python
print(f'{"variable":<18}{"value":>16}   dtype')
print('-' * 50)
for col in df.columns:
    print(f'{col:<18}{str(hit[col]):>16}   {df[col].dtype}')
```
