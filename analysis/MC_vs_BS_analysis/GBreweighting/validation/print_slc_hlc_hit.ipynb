# Print one SLC and one HLC hit from the same event
Source: `data_parquet/data_SplitInIcePulses_stopped.parquet`

```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

PARQUET = 'data_parquet/data_SplitInIcePulses_stopped.parquet'
df = pd.read_parquet(PARQUET)
print(f'Loaded {len(df):,} rows, {df["event_no"].nunique():,} unique events')
df.head()
```

```python
# Find an event that contains both an HLC hit (hlc==1) and an SLC hit (hlc==0)
has_hlc = df.loc[df['hlc'] == 1, 'event_no']
has_slc = df.loc[df['hlc'] == 0, 'event_no']
common = pd.Index(has_hlc.unique()).intersection(has_slc.unique())
event_no = int(common[0])
print(f'Using event_no = {event_no}')

event = df[df['event_no'] == event_no]
print(f'  total hits in event: {len(event)}  (HLC={int((event.hlc==1).sum())}, SLC={int((event.hlc==0).sum())})')
```

```python
hlc_hit = event[event['hlc'] == 1].iloc[[0]]
slc_hit = event[event['hlc'] == 0].iloc[[0]]

print(f'=== HLC hit (row index in parquet: {hlc_hit.index[0]}) ===')
print(hlc_hit.T)
print()
print(f'=== SLC hit (row index in parquet: {slc_hit.index[0]}) ===')
print(slc_hit.T)
```
